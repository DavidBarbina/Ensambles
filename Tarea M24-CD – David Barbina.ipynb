```python
# Librerías

import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
import os
from sklearn.datasets import load_iris

# Ajusta esta ruta si Graphviz está en otra carpeta
os.environ["PATH"] += os.pathsep + r"C:\Program Files\Graphviz\bin"

os.system("dot -V")  # Ahora debería devolver 0

df = pd.read_csv("https://drive.google.com/uc?export=download&id=1MlP9Lpb00Y4oJjGhixOGsa2k3twPY6si")
df.head()
```

```python
feature_cols = ["Age", "Sex", "BP", "Cholesterol","Na_to_K"]
X = df[feature_cols].values
y = df.Drug
```

```python
from sklearn import preprocessing

Cod_Sexo = preprocessing.LabelEncoder()
Cod_Sexo.fit(["F","M"])
X[:,1] = Cod_Sexo.transform(X[:,1])

Cod_BP = preprocessing.LabelEncoder()
Cod_BP.fit(["LOW", "NORMAL" , "HIGH"])
X[:,2] = Cod_BP.transform(X[:,2])

Cod_Colesterol = preprocessing.LabelEncoder()
Cod_Colesterol.fit(["NORMAL" , "HIGH"])
X[:,3] = Cod_Colesterol.transform(X[:,3])
y = df["Drug"].values
```

```python
from sklearn.ensemble import RandomForestClassifier

# Creación de grupos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state=1)

rf = RandomForestClassifier(n_estimators=100, random_state=1)
rf.fit(X_train, y_train)

# Estadísticas de desempeño generales

from sklearn.metrics import classification_report

y_pred = rf.predict(X_test)
print(classification_report(y_test, y_pred))

nuevos_datos = [
    [23, 0, 0, 0, 25.355],
    [47, 1, 1, 0, 13.093],
    [47, 1, 1, 0, 10.114],
    [28, 0, 2, 0, 7.798],
    [61, 0, 1, 0, 18.043],
    [22, 0, 2, 0, 8.607],
    [49, 0, 2, 0, 16.275],
    [41, 1, 1, 0, 11.037]
]

nuevos_datos = np.array(nuevos_datos, dtype=float)
prediccion = rf.predict(nuevos_datos)
print(prediccion)
```

```python
from sklearn.ensemble import GradientBoostingClassifier

# Creación de grupos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state=1)

gbt = GradientBoostingClassifier(n_estimators=100, random_state=1)  # 100 árboles de profundidad 3 (Default)
gbt.fit(X_train, y_train)

# Estadísticas de desempeño generales

y_pred = gbt.predict(X_test)
print(classification_report(y_test, y_pred))
```
